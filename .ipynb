```python
from pydantic import BaseModel, AnyUrl, ValidationError, field_validator

class ServerConfig(BaseModel):
    url: AnyUrl            # Pydantic crea un objeto `AnyUrl`, no un str

    # --- Pydantic 2: nuevo decorador ---
    @field_validator('url')          # puedes poner mode="before" o "after" si lo necesitas
    def url_debe_tener_puerto(cls, v: AnyUrl):  # v es la instancia `AnyUrl`
        if v.port is None:
            raise ValueError('La URL debe incluir un puerto, p. ej. https://dominio:443')
        return v

```

```python
# Prueba rápida
try:
    s = ServerConfig(url="https://ejemplo.com")
except ValidationError as e:
    print(e)

s_ok = ServerConfig(url="https://ejemplo.com:8443")
print(s_ok.url.port)
```
